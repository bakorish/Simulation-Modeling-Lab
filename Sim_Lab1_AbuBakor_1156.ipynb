```python
import numpy as np
```

```python
# Scalar
scalar = 3.7

# Array (6 elements)
arr = np.array([2.5, 4.8, 6.2, 8.9, 10.5, 12.3])

print("Scalar:", scalar)
print("Array:", arr)

# Trigonometric Functions
print("\n Trigonometric Functions ")
print("sin:", np.sin(arr))
print("cos:", np.cos(arr))
print("tan:", np.tan(arr))

# Clip values for inverse trigonometric functions
clip_arr = np.clip(arr / 15, -1, 1)

print("arcsin:", np.arcsin(clip_arr))
print("arccos:", np.arccos(clip_arr))
print("arctan:", np.arctan(arr))

# Hyperbolic Functions
print("\n  Hyperbolic Functions ")
print("sinh:", np.sinh(arr))
print("cosh:", np.cosh(arr))
print("tanh:", np.tanh(arr))

# Exponential & Logarithmic
print("\n Exponential & Logarithmic")
print("exp:", np.exp(arr))
print("log:", np.log(arr))
print("log10:", np.log10(arr))
print("log2:", np.log2(arr))

# Absolute & Square Root
print("\n Absolute & Square Root ")
print("absolute:", np.abs(arr))
print("sqrt:", np.sqrt(arr))

# Remainder
print("\n Remainder ")
print("arr % 2:", np.remainder(arr, 2))

# Rounding
print("\n Rounding ")
print("round:", np.round(arr))
print("floor:", np.floor(arr))
print("ceil:", np.ceil(arr))
```

```python
# Task 2: 3×2 Matrix Calculations
```

```python
import numpy as np

matrix = np.array([
    [1.5, 2.8],
    [3.2, 4.9],
    [5.6, 6.3]
])

print("Original Matrix")
print(matrix)


print("\nSin")
print(np.sin(matrix))

print("\nCos")
print(np.cos(matrix))

print("\nTan")
print(np.tan(matrix))


clip_matrix = np.clip(matrix / 7, -1, 1)

print("\nArcsin")
print(np.arcsin(clip_matrix))

print("\nArccos")
print(np.arccos(clip_matrix))

print("\nArctan")
print(np.arctan(matrix))



print("\nExponential")
print(np.exp(matrix))

print("\nNatural Log")
print(np.log(matrix))

print("\nLog10")
print(np.log10(matrix))

print("\nLog2")
print(np.log2(matrix))


print("\nAbsolute")
print(np.abs(matrix))

print("\nSquare Root")
print(np.sqrt(matrix))


print("\nRemainder when divided by 2")
print(np.remainder(matrix, 2))


print("\nRound")
print(np.round(matrix))

print("\nFloor")
print(np.floor(matrix))

print("\nCeil")
print(np.ceil(matrix))
```

```python
# Task 3: Vector Functions on a 2×3 Matrix
```

```python
import numpy as np

matrix = np.array([
    [2.1, 4.7, 6.3],
    [1.9, 8.5, 3.6]
])

print("Matrix")
print(matrix)

# Maximum & Minimum
print("\nMaximum Value:", np.max(matrix))
print("Maximum Index:", np.argmax(matrix))

print("\nMinimum Value:", np.min(matrix))
print("Minimum Index:", np.argmin(matrix))

# Vector Length
print("\nVector Length (Euclidean Norm):")
print(np.linalg.norm(matrix))

# Row-wise Descending Sort
print("\nRow-wise Descending Sort")
print(np.sort(matrix, axis=1)[:, ::-1])

# Entire Matrix Descending
print("\nEntire Matrix Descending")
print(np.sort(matrix.flatten())[::-1])

# Sum
print("\nSum")
print(np.sum(matrix))

# Product
print("\nProduct")
print(np.prod(matrix))

# Median
print("\nMedian")
print(np.median(matrix))

# Mean
print("\nMean")
print(np.mean(matrix))

# Standard Deviation
print("\nStandard Deviation")
print(np.std(matrix))
```

```python
# Heatmap Visualization
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

matrix = np.array([
    [2.1, 4.6, 6.4],
    [1.9, 8.5, 3.6]
])

plt.figure(figsize=(5,3))
sns.heatmap(matrix, annot=True, cmap="viridis")
plt.title("2×3 Matrix Heatmap")
plt.show()
```
